````python
# 验证现有数据集结构
def validate_dataset():
    """
    验证音频切片数据集的完整性
    """
    import os
    from pathlib import Path
    
    audio_dir = Path(AUDIO_DIR)
    text_dir = Path(TEXT_DIR)
    
    if not audio_dir.exists():
        print(f"❌ 音频目录不存在: {audio_dir}")
        return False
    
    if not text_dir.exists():
        print(f"❌ 文本目录不存在: {text_dir}")
        return False
    
    # 获取所有说话人文件夹
    audio_speakers = [d for d in audio_dir.iterdir() if d.is_dir()]
    text_speakers = [d for d in text_dir.iterdir() if d.is_dir()]
    
    print(f"✅ 找到 {len(audio_speakers)} 个音频说话人文件夹")
    print(f"✅ 找到 {len(text_speakers)} 个文本说话人文件夹")
    
    # 检查对应关系
    audio_speaker_names = {d.name for d in audio_speakers}
    text_speaker_names = {d.name for d in text_speakers}
    
    missing_text = audio_speaker_names - text_speaker_names
    missing_audio = text_speaker_names - audio_speaker_names
    
    if missing_text:
        print(f"⚠️  缺少文本的说话人: {missing_text}")
    if missing_audio:
        print(f"⚠️  缺少音频的说话人: {missing_audio}")
    
    # 统计文件数量
    total_audio_files = 0
    total_text_files = 0
    
    for speaker_dir in audio_speakers:
        audio_files = list(speaker_dir.glob("*.wav"))
        total_audio_files += len(audio_files)
    
    for speaker_dir in text_speakers:
        text_files = list(speaker_dir.glob("*.txt"))
        total_text_files += len(text_files)
    
    print(f"📊 总计音频文件: {total_audio_files}")
    print(f"📊 总计文本文件: {total_text_files}")
    
    return True

# 运行验证
validate_dataset()
````

````python
def load_wave(wave_path, sample_rate:int=16000) -> torch.Tensor:
    waveform, sr = torchaudio.load(wave_path, normalize=True)
    if sample_rate != sr:
        waveform = at.Resample(sr, sample_rate)(waveform)
    return waveform
````

````python
def get_italian_slice_audio_file_list(audio_dir, text_dir, text_max_length=120, audio_max_sample_length=480000, sample_rate=16000):
    """
    加载意大利语音频切片数据集
    数据结构：
    - audio_dir: 包含按说话人分组的音频切片文件
    - text_dir: 包含对应的文本转录文件
    """
    audio_transcript_pair_list = []
    audio_path = Path(audio_dir)
    text_path = Path(text_dir)
    
    if not audio_path.exists():
        print(f"音频目录 {audio_path} 不存在")
        return []
    
    if not text_path.exists():
        print(f"文本目录 {text_path} 不存在")
        return []
    
    # 获取所有说话人目录
    speaker_dirs = [d for d in audio_path.iterdir() if d.is_dir() and not d.name.startswith('.')]
    print(f"找到 {len(speaker_dirs)} 个说话人")
    
    for speaker_dir in tqdm(speaker_dirs, desc="处理说话人"):
        speaker_name = speaker_dir.name
        corresponding_text_dir = text_path / speaker_name
        
        if not corresponding_text_dir.exists():
            print(f"跳过说话人 {speaker_name}: 未找到对应的文本目录")
            continue
        
        # 获取该说话人的所有音频文件
        audio_files = list(speaker_dir.glob("*.wav"))
        
        for audio_file in audio_files:
            # 构建对应的文本文件路径
            text_file = corresponding_text_dir / f"{audio_file.stem}.txt"
            
            if not text_file.exists():
                continue
            
            try:
                # 读取文本内容
                with open(text_file, 'r', encoding='utf-8') as f:
                    text = f.read().strip()
                
                if not text:  # 跳过空文本
                    continue
                
                # 检查音频和文本长度
                audio = load_wave(audio_file, sample_rate=sample_rate)[0]
                if len(text) > text_max_length or len(audio) > audio_max_sample_length:
                    continue
                
                # 创建唯一的ID
                audio_id = f"{speaker_name}_{audio_file.stem}"
                audio_transcript_pair_list.append((audio_id, str(audio_file), text))
                
            except Exception as e:
                print(f"处理文件时出错 {audio_file}: {e}")
                continue
    
    return audio_transcript_pair_list

# 使用新的数据加载函数
print("正在加载意大利语音频切片数据集...")
all_audio_transcript_pairs = get_italian_slice_audio_file_list(
    AUDIO_DIR, TEXT_DIR, TEXT_MAX_LENGTH, AUDIO_MAX_LENGTH, SAMPLE_RATE
)

if len(all_audio_transcript_pairs) == 0:
    print("未找到数据！请检查数据集路径和格式。")
else:
    # 分割训练集和验证集
    train_num = int(len(all_audio_transcript_pairs) * TRAIN_RATE)
    train_audio_transcript_pair_list = all_audio_transcript_pairs[:train_num]
    eval_audio_transcript_pair_list = all_audio_transcript_pairs[train_num:]
    
    print(f"意大利语训练数据集数量: {len(train_audio_transcript_pair_list)}")
    print(f"意大利语验证数据集数量: {len(eval_audio_transcript_pair_list)}")
    
    # 显示一些示例
    print("\n数据示例:")
    for i, (audio_id, audio_path, text) in enumerate(all_audio_transcript_pairs[:3]):
        print(f"{i+1}. ID: {audio_id}")
        print(f"   文本: {text}")
        print(f"   音频: {Path(audio_path).name}")
        print()
````

````python
# 意大利语文本预处理函数
def text_italian_preprocess(text):
    """
    意大利语文本预处理
    对于意大利语，我们主要进行基本的清理
    """
    # 基本清理
    text = text.strip()
    # 移除多余空格
    import re
    text = re.sub(r'\s+', ' ', text)
    return text

print(text_italian_preprocess("Ciao, mi chiamo Marco Rossi."))
````

````python
# 配置Whisper为意大利语
woptions = whisper.DecodingOptions(language="it", without_timestamps=True)  # 改为意大利语
wmodel = whisper.load_model("base")
wtokenizer = whisper.tokenizer.get_tokenizer(True, language="it", task=woptions.task)  # 改为意大利语
````

````python
class ItalianSpeechDataset(torch.utils.data.Dataset):
    def __init__(self, audio_info_list, tokenizer, sample_rate) -> None:
        super().__init__()

        self.audio_info_list = audio_info_list
        self.sample_rate = sample_rate
        self.tokenizer = tokenizer

    def __len__(self):
        return len(self.audio_info_list)

    def __getitem__(self, id):
        audio_id, audio_path, text = self.audio_info_list[id]

        # 加载和处理音频
        audio = load_wave(audio_path, sample_rate=self.sample_rate)
        audio = whisper.pad_or_trim(audio.flatten())
        mel = whisper.log_mel_spectrogram(audio)

        # 处理意大利语文本
        text = text_italian_preprocess(text)
        text = [*self.tokenizer.sot_sequence_including_notimestamps] + self.tokenizer.encode(text)
        labels = text[1:] + [self.tokenizer.eot]

        return {
            "input_ids": mel,
            "labels": labels,
            "dec_input_ids": text
        }
````

````python
class WhisperDataCollatorWhithPadding:
    def __call__(self, features):  # 修正参数名从 sefl 到 self
        input_ids, labels, dec_input_ids = [], [], []
        for f in features:
            input_ids.append(f["input_ids"])
            labels.append(f["labels"])
            dec_input_ids.append(f["dec_input_ids"])

        input_ids = torch.concat([input_id[None, :] for input_id in input_ids])

        label_lengths = [len(lab) for lab in labels]
        dec_input_ids_length = [len(e) for e in dec_input_ids]
        max_label_len = max(label_lengths+dec_input_ids_length)

        labels = [np.pad(lab, (0, max_label_len - lab_len), 'constant', constant_values=-100) for lab, lab_len in zip(labels, label_lengths)]
        dec_input_ids = [np.pad(e, (0, max_label_len - e_len), 'constant', constant_values=50257) for e, e_len in zip(dec_input_ids, dec_input_ids_length)] # 50257 is eot token id

        batch = {
            "labels": labels,
            "dec_input_ids": dec_input_ids
        }

        batch = {k: torch.tensor(np.array(v), requires_grad=False) for k, v in batch.items()}
        batch["input_ids"] = input_ids

        return batch
````

````python
# 🛠️ 参数调节工具

def estimate_gpu_memory_usage(batch_size, audio_length_seconds=30, model_size="base"):
    """
    估算训练时的GPU内存使用量
    """
    # 基础内存使用 (MB)
    model_memory = {
        "tiny": 1000,
        "base": 2000, 
        "small": 3000,
        "medium": 5000,
        "large": 8000,
        "turbo": 5000
    }
    
    # 每个样本的大概内存使用 (MB)
    sample_memory = audio_length_seconds * 10  # 大概估算
    
    total_memory = model_memory[model_size] + batch_size * sample_memory
    
    print(f"📊 GPU内存使用估算:")
    print(f"  - 模型大小: {model_size} (~{model_memory[model_size]}MB)")
    print(f"  - 批处理大小: {batch_size}")
    print(f"  - 音频长度: {audio_length_seconds}秒")
    print(f"  - 预估总内存: ~{total_memory}MB ({total_memory/1024:.1f}GB)")
    print()
    
    # 建议
    if total_memory < 6000:
        print("✅ 适合6GB+GPU (如RTX 3060)")
    elif total_memory < 10000:
        print("⚠️  需要10GB+GPU (如RTX 3080)")
    elif total_memory < 16000:
        print("🔥 需要16GB+GPU (如RTX 4080)")
    else:
        print("💰 需要24GB+GPU (如RTX 4090)")

def suggest_parameters(gpu_memory_gb, data_size, target_quality="medium"):
    """
    根据GPU内存和数据量建议参数
    """
    print(f"🎯 针对 {gpu_memory_gb}GB GPU 的参数建议:")
    print()
    
    # 根据GPU内存建议batch size
    if gpu_memory_gb <= 8:
        batch_size = 2
        accumulation = 4
    elif gpu_memory_gb <= 12:
        batch_size = 4
        accumulation = 2
    elif gpu_memory_gb <= 16:
        batch_size = 8
        accumulation = 1
    else:
        batch_size = 16
        accumulation = 1
    
    # 根据数据量建议训练轮数
    if data_size < 1000:
        epochs = 20
    elif data_size < 5000:
        epochs = 15
    else:
        epochs = 10
    
    # 根据质量要求建议学习率
    lr_map = {
        "fast": 0.001,
        "medium": 0.0005,
        "high": 0.0001
    }
    
    print(f"推荐配置:")
    print(f"  batch_size = {batch_size}")
    print(f"  gradient_accumulation_steps = {accumulation}")
    print(f"  num_train_epochs = {epochs}")
    print(f"  learning_rate = {lr_map[target_quality]}")
    print(f"  实际批次大小 = {batch_size * accumulation}")

# 示例使用
print("=== 参数调节工具演示 ===")
estimate_gpu_memory_usage(batch_size=4, audio_length_seconds=30, model_size="base")
suggest_parameters(gpu_memory_gb=8, data_size=3000, target_quality="medium")
````

````python
class WhisperModelModule(LightningModule):
    def __init__(self, cfg:Config, model_name="base", lang="it", train_dataset=[], eval_dataset=[]) -> None:
        super().__init__()
        self.options = whisper.DecodingOptions(language=lang, without_timestamps=True)
        self.model = whisper.load_model(model_name)
        self.tokenizer = whisper.tokenizer.get_tokenizer(True, language=lang, task=self.options.task)

        # 只训练解码器
        for p in self.model.encoder.parameters():
            p.requires_grad = False

        self.loss_fn = nn.CrossEntropyLoss(ignore_index=-100)
        self.metrics_wer = evaluate.load("wer")
        self.metrics_cer = evaluate.load("cer")

        self.cfg = cfg
        self.__train_dataset = train_dataset
        self.__eval_dataset = eval_dataset

    def forward(self, x):
        return self.model(x)

    def training_step(self, batch, batch_id):
        input_ids = batch["input_ids"]
        labels = batch["labels"].long()
        dec_input_ids = batch["dec_input_ids"].long()

        with torch.no_grad():
            audio_features = self.model.encoder(input_ids)

        out = self.model.decoder(dec_input_ids, audio_features)
        loss = self.loss_fn(out.view(-1, out.size(-1)), labels.view(-1))
        self.log("train/loss", loss, on_step=True, prog_bar=True, logger=True)
        return loss

    def validation_step(self, batch, batch_id):
        input_ids = batch["input_ids"]
        labels = batch["labels"].long()
        dec_input_ids = batch["dec_input_ids"].long()

        audio_features = self.model.encoder(input_ids)
        out = self.model.decoder(dec_input_ids, audio_features)

        loss = self.loss_fn(out.view(-1, out.size(-1)), labels.view(-1))

        out[out == -100] = self.tokenizer.eot
        labels[labels == -100] = self.tokenizer.eot

        o_list, l_list = [], []
        for o, l in zip(out, labels):
            o = torch.argmax(o, dim=1)
            o_list.append(self.tokenizer.decode(o, skip_special_tokens=True))
            l_list.append(self.tokenizer.decode(l, skip_special_tokens=True))
        cer = self.metrics_cer.compute(references=l_list, predictions=o_list)
        wer = self.metrics_wer.compute(references=l_list, predictions=o_list)

        self.log("val/loss", loss, on_step=True, prog_bar=True, logger=True)
        self.log("val/cer", cer, on_step=True, prog_bar=True, logger=True)
        self.log("val/wer", wer, on_step=True, prog_bar=True, logger=True)

        return {
            "cer": cer,
            "wer": wer,
            "loss": loss
        }

    def configure_optimizers(self):
        """优化器和调度器配置"""
        model = self.model
        no_decay = ["bias", "LayerNorm.weight"]
        optimizer_grouped_parameters = [
            {
                "params": [p for n, p in model.named_parameters()
                            if not any(nd in n for nd in no_decay)],
                "weight_decay": self.cfg.weight_decay,
            },
            {
                "params": [p for n, p in model.named_parameters()
                            if any(nd in n for nd in no_decay)],
                "weight_decay": 0.0,
            },
        ]
        optimizer = AdamW(optimizer_grouped_parameters,
                          lr=self.cfg.learning_rate,
                          eps=self.cfg.adam_epsilon)
        self.optimizer = optimizer

        scheduler = get_linear_schedule_with_warmup(
            optimizer, num_warmup_steps=self.cfg.warmup_steps,
            num_training_steps=self.t_total
        )
        self.scheduler = scheduler

        return [optimizer], [{"scheduler": scheduler, "interval": "step", "frequency": 1}]

    def setup(self, stage=None):
        """初始设置"""
        if stage == 'fit' or stage is None:
            self.t_total = (
                (len(self.__train_dataset) // (self.cfg.batch_size))
                // self.cfg.gradient_accumulation_steps
                * float(self.cfg.num_train_epochs)
            )

    def train_dataloader(self):
        """训练数据加载器"""
        dataset = ItalianSpeechDataset(self.__train_dataset, self.tokenizer, self.cfg.sample_rate)
        return torch.utils.data.DataLoader(dataset,
                          batch_size=self.cfg.batch_size,
                          drop_last=True, shuffle=True, num_workers=self.cfg.num_worker,
                          collate_fn=WhisperDataCollatorWhithPadding()
                          )

    def val_dataloader(self):
        """验证数据加载器"""
        dataset = ItalianSpeechDataset(self.__eval_dataset, self.tokenizer, self.cfg.sample_rate)
        return torch.utils.data.DataLoader(dataset,
                          batch_size=self.cfg.batch_size,
                          num_workers=self.cfg.num_worker,
                          collate_fn=WhisperDataCollatorWhithPadding()
                          )
````
